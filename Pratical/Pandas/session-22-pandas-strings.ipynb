```python
# import titanic

df = pd.read_csv('dataset_files\session-22\\titanic.csv')
```

```python
df.head()
```

```python
df['Name']
```

```python
# Common Functions
# lower/upper/capitalize/title
df['Name'].str.upper() # upper 
df['Name'].str.capitalize() # first letter capital
df['Name'].str.title() # first letter capital of every word

# len
df['Name'][df['Name'].str.len().max()]

# strip
df['Name'].str.strip() # use to remove all unnecessary spaces
```

```python
# split -> get
df['lastname'] = df['Name'].str.split(',').str.get(0)
```

```python
df.head()
```

```python
df[['title','firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace
df['title'] = df['title'].str.replace('Mrs.','Miss')
df['title'] = df['title'].str.replace('Mlle','Miss')
```

```python
df['title'].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['Name'].str.endswith('A')]

```

```python
# isdigit/isalpha
df[df['Name'].str.isdigit()]
df[df['Name'].str.isalpha()]
```

```python
# applying regex
# contains
# search john -> both case
df[df['firstname'].str.contains('john',case=False)]

# find lastnames with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')] # here ^ is startswith and $ is endswiths
```

```python
df[df['lastname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')] # here ^ is startswith and $ is endswiths
# ^ inside bracket act as not
```

```python
# slicing

df['Name'].str[1:4]
df['Name'].str[::2]
df['Name'].str[::-1]
```
