Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

```python
import numpy as np
import pandas as pd
```

# Series from list


```python
# Series from list

country = ["India","USA","Nepal","China","UAE","Japan","USSR"]

print(pd.Series(country)) # In series str are 99% recognize as obj
```

```python
# Integers

runs = [12,55,67,89,100]

runs_ser = pd.Series(runs)
print(runs_ser)
```

```python
runs
```

```python
# custom index

marks = [57,77,89,100]
subjects = ['English','Science','History','Maths']

pd.Series(marks,index=subjects)
```

```python
# setting a name

marks = pd.Series(marks,index=subjects, name='Himanshu ke marks')
print(marks)
```

# Series from dict

```python
marks = {
    'English': 67,
    'Science': 89,
    'History': 77,
    'Maths':100
}

marks_series = pd.Series(marks, name='Himanshu ke marks')
print(marks_series)
```

Series with String Index
1. No ambiguity: If your Series has string labels (e.g., 'a', 'b', 'c'), it is completely impossible for -1 to be a label.
2. Automatic fallback: Pandas recognizes that -1 is an integer and safely falls back to standard Python positional indexing (meaning "the last element").
