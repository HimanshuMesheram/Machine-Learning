```python
# legend -> location
plt.plot(batsman['index'],batsman['V Kohli'], color='pink', linestyle='dotted', linewidth=3,marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'], linestyle='dashdot',linewidth=3, marker='o', label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc='best')
```

```python
# limiting axes
price=[48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020, 2021]

plt.plot(year,price)
plt.ylim(0,75000)
plt.xlim(2017,2019)
```

# Scatter Plot

- Bivariate Analysis
- numerical vs numerical
- Use case - Finding correlation

```python
# plt.scatter

x = np.linspace(-10,10,50)

y = 10*x + 10 + np.random.randint(0,200,50)

plt.scatter(x,y)
```

```python
# plt.scatter on pandas data
df = pd.read_csv('data_sample\\batter.csv')
df

df = df.head(50)
df
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red', marker='^')

plt.title('Average and Strike Rate of Top 50 batsman')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# marker
```

```python
# size

tips = sns.load_dataset('tips')

plt.scatter(tips['total_bill'],tips['tip'], s=tips['size']*20)
```

```python
# scatterplot using plt.plot
# plot is faster as compare to scatter

plt.plot(tips['total_bill'],tips['tip'],'o') # just use marker like 'o'
```

# Bar Chart

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [20,30,10,17,25]
colors = ['red','blue','yellow','green','pink']

plt.bar(colors,children, color='orange')
```

```python
# bar chart using data
```

```python
# horizontal bar chart
plt.barh(colors,children, color='orange')
```

```python
# color and label
df = pd.read_csv('data_sample\\batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
# Multiple Bar charts
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2,df['2016'],width=0.2)

plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show()
```

```python
# name problem
children = [20,30,10,17,25]
colors = ['red redredredredredred','blue blueblueblueblueblue','yellowyellowyellowyellow','greengreengreengreengreen','pinkpinkpinkpinkpinkpink']

plt.bar(colors,children, color='orange')
plt.xticks(rotation='vertical')
```

```python
# Stacked Bar chart
plt.bar(df['batsman'],df['2017'], label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

# Pie Chart

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data

data = [20,35,100,56,89]
subjects = ['Hindi','Physic','Maths','Science','Computer']

plt.pie(data, labels=subjects)
plt.show()
```

```python
# dataset
df = pd.read_csv('data_sample\gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

```python
# percentage and colors
plt.pie(df['batsman_runs'],labels=df['batsman'], autopct='%0.1f%%', colors=['red','orange','green','yellow','pink','cyan'])
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'],labels=df['batsman'], autopct='%0.1f%%',explode=[0.3,0,0,0.3,0,0],shadow=True)
plt.show()
```
