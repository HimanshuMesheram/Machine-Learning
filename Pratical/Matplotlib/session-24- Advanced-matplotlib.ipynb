# Annotation

```python
batter = pd.read_csv('data_sample\\batter.csv')
batter
```

```python
sample_df = batter.head(100).sample(25,random_state=5)
sample_df
```

```python
plt.figure(figsize=(18,9))

plt.scatter(sample_df['avg'],sample_df['strike_rate'], s=sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

# Horizontal and Vertical line

```python
plt.figure(figsize=(18,9))

plt.scatter(sample_df['avg'],sample_df['strike_rate'], s=sample_df['runs'])

plt.axhline(135,color='red')
plt.axvline(30,color='red')
plt.axhline(140,color='green')


for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
# A diff way to plot graphs
plt.figure(figsize=(15,6))
plt.scatter(batter['avg'],batter['strike_rate'])
plt.title('Something')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
```

```python
fig,ax = plt.subplots(figsize=(15,6))

ax.scatter(batter['avg'],batter['strike_rate'],color='red',marker='+')
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')

fig.show()
```

```python
# dataset
fig , ax = plt.subplots(nrows=2,ncols=1, sharex=True, figsize=(10,7))

ax[0].scatter(batter['avg'],batter['strike_rate'])
ax[1].scatter(batter['avg'],batter['runs'])

ax[0].set_title('Avg vs Strike rate')
ax[0].set_ylabel('strike rate')

ax[1].set_title('Avg vs Runs')
ax[1].set_ylabel('runs')
ax[1].set_xlabel('avg')
```

```python
fig , ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batter['avg'],batter['strike_rate'])
ax[0,1].scatter(batter['avg'],batter['runs'])
ax[1,0].hist(batter['avg'])
ax[1,1].hist(batter['runs'])

ax[0,0].set_title('Avg vs Strike rate')
ax[0,0].set_ylabel('strike rate')

ax[0,1].set_title('Avg vs Runs')
ax[0,1].set_ylabel('runs')
# ax[1].set_xlabel('avg')
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])

ax4 = fig.add_subplot(2,2,4)
ax4.hist(batter['strike_rate'])
```

# 3D Surface plot

```python


x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
xx.shape
yy.shape
```

```python
z = xx**2 + yy**2
```

```python
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))

z = np.sin(xx) + np.cos(yy)

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# Contour plot

```python
fig = plt.figure(figsize=(12,8))

z = np.sin(xx) + np.cos(yy)

ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contour fillgraph

fig = plt.figure(figsize=(12,8))

z = np.sin(xx) + np.cos(yy)

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```
