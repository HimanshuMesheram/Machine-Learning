# Data Cleaning and Data Preprocessing

```python
df_cleaned = df.copy()
```

```python
df_cleaned.head()
```

```python
df_cleaned['trans_date_trans_time'] = pd.to_datetime(df_cleaned['trans_date_trans_time'])
```

```python
df_cleaned["trans_hour"] = df_cleaned["trans_date_trans_time"].dt.hour

df_cleaned["trans_day"] = df_cleaned["trans_date_trans_time"].dt.day

df_cleaned["trans_month"] = df_cleaned["trans_date_trans_time"].dt.month

df_cleaned["trans_dayofweek"] = df_cleaned["trans_date_trans_time"].dt.dayofweek
```

```python
# df_cleaned.drop("trans_date_trans_time", axis=1, inplace=True)
```

```python
df_cleaned['dob'] = pd.to_datetime(df_cleaned['dob'])
```

```python
df_cleaned['dob'] = pd.to_datetime(df_cleaned['dob'])
```

```python
df_cleaned['age'] = (
    (df_cleaned['trans_date_trans_time'] - df_cleaned['dob']).dt.days / 365.25
)
```

```python
X = df.drop(columns=["is_fraud"])
```

```python
df_cleaned.drop("trans_num", axis=1, inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression 
from sklearn.compose import ColumnTransformer

import xgboost as xgb
```

```python
numeric_features = [
    "amt",
    "lat",
    "long",
    "city_pop",
    "merch_lat",
    "merch_long",
    "trans_hour",
    "trans_day",
    "trans_month",
    "trans_dayofweek",
    "age"
]
```

```python
categorical_features = [
    "merchant",
    "category",
    "city",
    "state",
    "job"
]
```

```python
print(df["merchant"].nunique())
print(df["city"].nunique())
print(df["job"].nunique())
```

```python
X = df_cleaned.drop(columns=['is_fraud'])
y = df_cleaned['is_fraud']
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size=0.33, random_state=42,stratify=y)     
```

```python
ct = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric_features),
    ('cat', OneHotEncoder(handle_unknown="ignore"), categorical_features)   
])
```

```python
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
```

```python
pipeline = Pipeline([
    ('Col_transformer',ct),
    ('Smote',SMOTE(random_state=42)),
    # ('logistic regression',LogisticRegression(max_iter=1000,random_state=42))
    ('xgboost', xgb.XGBClassifier(random_state=42, eval_metric='logloss'))
])
```

```python
pipeline.fit(X_train,y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

# Evaluation

```python
# Logistic

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
# XGB

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
y_prob = pipeline.predict_proba(X_test)[:, 1]
```

# ROC-AUC

```python
print(
    "ROC-AUC:",
    roc_auc_score(y_test, y_prob)
)
```

# PR-AUC

```python
print(
    "PR-AUC:",
    average_precision_score(
        y_test,
        y_prob
    )
)
```

# CV

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring="f1",
    n_jobs=-1
)

print("CV F1 Scores:", scores)
print("Mean CV F1:", scores.mean())
```

# GridSearchCV

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "xgboost__n_estimators": [100, 200],
    "xgboost__max_depth": [3, 5, 7],
    "xgboost__learning_rate": [0.05, 0.1],
    "xgboost__subsample": [0.8, 1.0]
}

grid = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)

print("Best Parameters:")
print(grid.best_params_)

print("Best CV F1:")
print(grid.best_score_)
```
