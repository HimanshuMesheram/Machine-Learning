**Q31.**  
Compute the **Z-score** of `np.array([10, 20, 30, 40, 50])` manually:
- Subtract the mean
- Divide by the standard deviation

Use `np.mean` and `np.std`. Verify the Z-scored array has mean ≈ 0 and std ≈ 1.


````python
# Create the original array
arr = np.array([10, 20, 30, 40, 50])

# Calculate the mean of the array
mean = np.mean(arr)

# Calculate the standard deviation of the array
std = np.std(arr)

# Compute the Z-score for each element
# Formula: Z = (x - mean) / std
z_score = (arr - mean) / std

# Print the original array
print("Original Array:")
print(arr)

# Print mean and standard deviation
print("\nMean:", mean)
print("Standard Deviation:", std)

# Print the Z-score array
print("\nZ-score Array:")
print(z_score)

# Verify that the Z-score array has mean approximately 0
print("\nMean of Z-score Array:", np.mean(z_score))

# Verify that the Z-score array has standard deviation approximately 1
print("Standard Deviation of Z-score Array:", np.std(z_score))
````

````python
# Ill-scaled dataset
X = np.random.rand(50, 4) * np.array(
    [1, 100, 1000, 0.01]
)

# -----------------------------------
# Step 1: Min-Max Scaling
# -----------------------------------
col_min = X.min(axis=0)

col_max = X.max(axis=0)

X_minmax = (
    (X - col_min)
    /
    (col_max - col_min)
)

print("Min Values:")
print(X_minmax.min(axis=0))

print("\nMax Values:")
print(X_minmax.max(axis=0))

# -----------------------------------
# Step 2: Z-score
# -----------------------------------
mean = X_minmax.mean(axis=0)

std = X_minmax.std(axis=0)

# Safe handling for std = 0
safe_std = np.where(
    std == 0,
    1,
    std
)

X_z = (
    X_minmax - mean
) / safe_std

print("\nMeans After Z-score:")
print(X_z.mean(axis=0))

print("\nStds After Z-score:")
print(X_z.std(axis=0))
````

**Q100.**  
**Full ML pipeline challenge** — implement everything from scratch using only Sections 1–13:

```python
X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)  # binary target
```

1. **Z-score standardize** `X` column-wise using broadcasting
2. **Shuffle and split** 70/15/15 using `np.argsort(np.random.rand(200))` as shuffle index
3. **Initialize weights** `w = np.zeros(5)` and bias `b = 0.0`
4. **Forward pass** on `X_train`: `z = X_train @ w + b`, `y_hat = 1 / (1 + np.exp(-z))`
5. **BCE loss**: clip `y_hat` to `[1e-7, 1-1e-7]`, compute `-np.mean(y*log + (1-y)*log)`
6. **Gradient**: `grad_w = X_train.T @ (y_hat - y_train) / len(y_train)`
7. **One gradient descent step**: `w -= 0.1 * grad_w`
8. **Forward pass again** with updated `w`, compute new BCE loss
9. Print: initial loss, updated loss (should be slightly lower), gradient norm `np.sqrt((grad_w**2).sum())`

````python
import numpy as np

# -----------------------------------
# Create dataset
# -----------------------------------
X = np.random.rand(200, 5)

y = (
    (X[:, 0] + X[:, 1]) > 1.0
).astype(int)

# -----------------------------------
# Step 1: Z-score standardization
# -----------------------------------
mean = X.mean(axis=0)

std = X.std(axis=0)

X = (X - mean) / std

# -----------------------------------
# Step 2: Shuffle and split
# -----------------------------------
shuffle_idx = np.argsort(
    np.random.rand(200)
)

X = X[shuffle_idx]
y = y[shuffle_idx]

n = len(X)

train_end = int(0.70 * n)
val_end = int(0.85 * n)

X_train = X[:train_end]
y_train = y[:train_end]

X_val = X[train_end:val_end]
y_val = y[train_end:val_end]

X_test = X[val_end:]
y_test = y[val_end:]

print("Train Shape:", X_train.shape)
print("Val Shape  :", X_val.shape)
print("Test Shape :", X_test.shape)

# -----------------------------------
# Step 3: Initialize parameters
# -----------------------------------
w = np.zeros(5)

b = 0.0

# -----------------------------------
# Step 4: Forward Pass
# -----------------------------------
z = X_train @ w + b

y_hat = 1 / (
    1 + np.exp(-z)
)

# -----------------------------------
# Step 5: BCE Loss
# -----------------------------------
y_hat_clip = np.clip(
    y_hat,
    1e-7,
    1 - 1e-7
)

initial_loss = -np.mean(
    y_train * np.log(y_hat_clip)
    +
    (1 - y_train)
    * np.log(1 - y_hat_clip)
)

# -----------------------------------
# Step 6: Gradient
# -----------------------------------
grad_w = (
    X_train.T
    @
    (y_hat - y_train)
) / len(y_train)

# -----------------------------------
# Step 7: Gradient Descent
# -----------------------------------
learning_rate = 0.1

w -= learning_rate * grad_w

# -----------------------------------
# Step 8: Forward Again
# -----------------------------------
z2 = X_train @ w + b

y_hat2 = 1 / (
    1 + np.exp(-z2)
)

y_hat2 = np.clip(
    y_hat2,
    1e-7,
    1 - 1e-7
)

updated_loss = -np.mean(
    y_train * np.log(y_hat2)
    +
    (1 - y_train)
    * np.log(1 - y_hat2)
)

# -----------------------------------
# Step 9: Gradient Norm
# -----------------------------------
grad_norm = np.sqrt(
    (grad_w ** 2).sum()
)

print("\nInitial Loss:")
print(initial_loss)

print("\nUpdated Loss:")
print(updated_loss)

print("\nGradient Norm:")
print(grad_norm)

print("\nLoss Improved:")
print(updated_loss < initial_loss)
````
