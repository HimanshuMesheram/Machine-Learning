36. Extract the integer part of a random array of positive numbers using 4 different methods (★★☆) `hint: %, np.floor, astype, np.trunc`

````python
# Generate an array of random positive floating-point numbers
Z = np.random.uniform(0, 10, 10)

# Display original values
print("Original Array:")
print(Z)

# Method 1:
# Convert float values to integers by removing decimal part
R1 = Z.astype(int)

# Method 2:
# Take the floor value of each element
R2 = np.floor(Z)

# Method 3:
# Truncate (remove) the fractional part
R3 = np.trunc(Z)

# Method 4:
# Remove the fractional part using modulus operation
R4 = Z - (Z % 1)

# Display all results
print("\nUsing astype(int):")
print(R1)

print("\nUsing np.floor():")
print(R2)

print("\nUsing np.trunc():")
print(R3)

print("\nUsing modulus (%):")
print(R4)
````

38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)
`hint: np.fromiter`

````python
# Generator function
# It generates integers from 0 to 9 one by one
def generate_numbers():
    
    # Loop 10 times
    for i in range(10):
        
        # Yield returns one value at a time
        yield i

# Create a NumPy array from the generator
# dtype specifies the data type of elements
Z = np.fromiter(generate_numbers(), dtype=int)

# Display the array
print(Z)
````

44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)
`hint: np.sqrt, np.arctan2`

````python
# Create a random 10x2 matrix
# Each row represents a point (x, y)
Z = np.random.random((10, 2))

# Display original Cartesian coordinates
print("Cartesian Coordinates:")
print(Z)

# Extract x coordinates (first column)
X = Z[:, 0]

# Extract y coordinates (second column)
Y = Z[:, 1]

# Compute distance from origin
# r = sqrt(x² + y²)
R = np.sqrt(X**2 + Y**2)

# Compute angle in radians
# arctan2 handles all quadrants correctly
Theta = np.arctan2(Y, X)

# Combine radius and angle into one matrix
polar = np.column_stack((R, Theta))

# Display polar coordinates
print("\nPolar Coordinates (r, θ):")
print(polar)
````

47. Given two arrays, X and Y, construct the Cauchy matrix C (Cij =1/(xi - yj)) (★★☆)
`hint: np.subtract.outer`

````python
# Create two arrays
X = np.array([1, 2, 3, 4])
Y = np.array([5, 6, 7])

# Compute all possible differences (xi - yj)
# subtract.outer() creates a matrix where:
# element (i,j) = X[i] - Y[j]
D = np.subtract.outer(X, Y)

# Compute the Cauchy matrix
# Cij = 1 / (xi - yj)
C = 1.0 / D

# Display the result
print("Cauchy Matrix:")
print(C)
````

52. Consider a random vector with shape (100,2) representing coordinates, find point by point distances (★★☆)
`hint: np.atleast_2d, T, np.sqrt`

````python
arr = np.random.random((10,2))

R, C = np.atleast_2d(arr[:,0],arr[:,1])

dis = np.sqrt((R - R.T) ** 2 + (C - C.T ** 2)) # Used broadcasting

print("Distance:",dis)
````

# How This Works Step-by-Step
1. np.atleast_2d(Z[:, 0], Z[:, 1]): Slicing 1D columns Z[:, 0] and Z[:, 1] normally yields a 1D shape (100,). This function elevates them to 2D row vectors with a shape of (1, 100).
2. X.T and Y.T: Transposing the row vectors turns them into column vectors with a shape of (100, 1).
3. Broadcasting (X - X.T): When you subtract a (100, 1) column vector from a (1, 100) row vector, NumPy automatically stretches both matrices into a (100, 100) grid. This evaluates the difference between every single combination of coordinates (\(x_i - x_j\)).
4. np.sqrt: Computes the element-wise square root of the combined squared differences to complete the standard Euclidean distance equation: \(\sqrt{(x_{i}-x_{j})^{2}+(y_{i}-y_{j})^{2}}\).

56. Generate a generic 2D Gaussian-like array (★★☆)

80. Consider an arbitrary array, write a function that extracts a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)
`hint: minimum maximum`

````python
def extract(arr, center, shape, fill=0):

    # Create output array filled with fill value
    result = np.full(shape, fill)

    r, c = center
    h, w = shape

    # Source boundaries inside original array
    r0 = max(r - h//2, 0)
    c0 = max(c - w//2, 0)

    r1 = min(r + h//2 + 1, arr.shape[0])
    c1 = min(c + w//2 + 1, arr.shape[1])

    # Destination boundaries inside result
    rr0 = max(0, h//2 - r)
    cc0 = max(0, w//2 - c)

    # Copy valid data
    result[
        rr0:rr0+(r1-r0),
        cc0:cc0+(c1-c0)
    ] = arr[r0:r1, c0:c1]

    return result

# Example array
A = np.arange(25).reshape(5,5)

# Extract 3x3 centered at top-left corner
print(extract(A, (0,0), (3,3), fill=-1))
````

85. Create a 2D array subclass such that Z[i,j] == Z[j,i] (★★★)
`hint: class method`

````python
# Create a custom NumPy array class
class SymmetricArray(np.ndarray):

    # Create the object
    def __new__(cls, shape):

        # Create an array of zeros
        obj = np.zeros(shape).view(cls)

        return obj

    # Override assignment operation
    def __setitem__(self, index, value):

        i, j = index

        # Assign value to (i,j)
        super().__setitem__((i, j), value)

        # Assign same value to (j,i)
        super().__setitem__((j, i), value)

# Create a 4x4 symmetric matrix
Z = SymmetricArray((4,4))

# Assign value
Z[1,3] = 10

print(Z)
````

88. How to implement the Game of Life using numpy arrays? (★★★)
`No hints provided...`

````python
def game_of_life(Z):

    # Count neighbors
    N = sum(
        np.roll(np.roll(Z, i, axis=0), j, axis=1)
        for i in (-1,0,1)
        for j in (-1,0,1)
        if not (i == 0 and j == 0)
    )

    # Birth condition
    birth = (N == 3) & (Z == 0)

    # Survival condition
    survive = ((N == 2) | (N == 3)) & (Z == 1)

    # Reset board
    Z[...] = 0

    # Apply rules
    Z[birth | survive] = 1

    return Z

# Random initial board
Z = np.random.randint(0,2,(10,10))

print(game_of_life(Z))
````

98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
`hint: np.cumsum, np.interp`

````python
X = np.array([0,1,2,4])
Y = np.array([0,1,1,2])

# Distance between consecutive points
dist = np.sqrt(
    np.diff(X)**2 +
    np.diff(Y)**2
)

# Cumulative distance
cumdist = np.insert(
              np.cumsum(dist),
              0,
              0
          )

# Create equally spaced distances
samples = np.linspace(
              0,
              cumdist[-1],
              10
          )

# Interpolate coordinates
Xs = np.interp(samples, cumdist, X)
Ys = np.interp(samples, cumdist, Y)

print(Xs)
print(Ys)
````

99. Given an integer n and a 2D array X, select from X the rows which can be interpreted as draws from a multinomial distribution with n degrees, i.e., the rows which only contain integers and which sum to n. (★★★)
`hint: np.logical_and.reduce, np.mod`

````python
n = 5

X = np.array([
    [2,2,1],
    [1,2,1],
    [3,1,1],
    [2.5,1,1]
])

# Check integer condition
integer_check = np.mod(X,1) == 0

# Check sum condition
sum_check = X.sum(axis=1) == n

# Combine both conditions
mask = np.logical_and.reduce(
           (
               integer_check.all(axis=1),
               sum_check
           )
       )

print(X[mask])
````

100. Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means). (★★★)
`hint: np.percentile`

````python
# Original data
X = np.random.randn(100)

# Number of bootstrap samples
N = 10000

# Generate bootstrap samples
samples = np.random.choice(
              X,
              (N, len(X)),
              replace=True
          )

# Mean of each bootstrap sample
means = samples.mean(axis=1)

# 95% confidence interval
CI = np.percentile(
         means,
         [2.5, 97.5]
     )

print("95% Confidence Interval:")
print(CI)
````
