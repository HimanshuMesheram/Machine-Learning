# Salary Distribution Across Job Titles

```python
salary_job = emp.groupby('JobTitle')['Salary'].agg(AverageSalary='mean',MaximunSalary='max',MinimumSalary='min')
salary_job
```

Insight
- The highest-paid job role is HR Manager and lowest-paid job roles is HR Manager every other job role is average in salary.
- Higher salaries generally correspond to senior or specialized positions.

# Salary Distribution by Gender

```python
gender_salary = emp.groupby('Gender')['Salary'].mean()
gender_salary
```

Insight
- Compare average salaries of male and female employees we can identify that female earn more as compare to male.
- Detect minor gender pay gap.

# Salary based on location

```python
location_salary = emp.groupby("Location")["Salary"].mean().sort_values(ascending=False)

print(location_salary)
```

Insight
- Location-wise Seattle and San Francisco have the highest salaries.
- High-cost cities generally pay more.
