## NP41. Create a synthetic ML dataset using NumPy.
> 💡 **Hint:** Use `np.random.normal`, `np.random.randint`, `np.random.choice` to generate features. Use `np.where` or `np.select` to create a binary/multi-class label column.


```python
# Set seed for reproducibility
np.random.seed(42)
n_samples = 1000

# 1. Generate Features using NumPy
# Continuous features (e.g., Age, Income)
age = np.random.randint(18, 70, size=n_samples)
income = np.random.normal(loc=50000, scale=15000, size=n_samples).round(2)

# Categorical feature (e.g., Education Level)
edu_levels = ["High School", "Bachelor", "Master", "PhD"]
education = np.random.choice(edu_levels, size=n_samples, p=[0.3, 0.4, 0.2, 0.1])

# 2. Define logic for the Target Label using np.select
# Let's create a multi-class label: Risk Category (Low, Medium, High)
# Based on conditions involving Age and Income
conditions = [
    (income > 65000) & (age > 30),  # High income, stable age -> Low Risk
    (income < 35000) | (age < 25),  # Low income or very young -> High Risk
]
choices = ["Low", "High"]

# np.select applies conditions; default value handles anything that doesn't match
risk_label = np.select(conditions, choices, default="Medium")

# 3. Combine into a structured Pandas DataFrame
dataset = pd.DataFrame(
    {
        "Age": age,
        "Income": income,
        "Education": education,
        "Risk_Rating": risk_label,
    }
)

# Preview the synthetic dataset
print(dataset.head())
```

## NP42. Detect and replace outliers using NumPy.
> 💡 **Hint:** Compute Q1, Q3 with `np.percentile`. Define bounds as `Q1 - 1.5*IQR` and `Q3 + 1.5*IQR`. Use `np.clip(arr, lo, hi)` to replace outliers with boundary values.


```python
arr = np.array([12, 15, 14, 1000, 19, 11, 16, -500, 14, 18, 15])

q1 = np.percentile(arr,25)
q3 = np.percentile(arr,75)

iqr = q3 - q1

lower_bo = q1 - 1.5 * iqr
upper_bo = q3 + 1.5 * iqr

clean_arr = np.clip(arr,lower_bo,upper_bo)

clean_arr
```

### NP43. Scale dataset using NumPy functions.
> 💡 **Hint:** Apply Min-Max or Z-score scaling using NumPy on the `.values` array. Store results as new columns in the DataFrame.

```python
# 1. Create a sample dataset
df = pd.DataFrame({"salary": [30000, 50000, 85000, 120000, 45000]})

# Extract the underlying NumPy array
X = df["salary"].values

# 2. Min-Max Scaling (Scales data between 0 and 1)
x_min = np.min(X)
x_max = np.max(X)
df["salary_minmax"] = (X - x_min) / (x_max - x_min)

# 3. Z-score Scaling (Scales data to mean = 0, standard deviation = 1)
x_mean = np.mean(X)
x_std = np.std(X)
df["salary_zscore"] = (X - x_mean) / x_std

# View the final DataFrame
print(df)
```

## 🎯 Quick Hints Reference

| Task | NumPy Function | Pandas Equivalent |
|------|---------------|-------------------|
| Convert to array | `arr = df["col"].values` | `arr = df["col"].to_numpy()` |
| Conditional assign | `np.where(cond, a, b)` | `df.loc[cond, "col"] = val` |
| Multi-condition | `np.select([c1,c2],[v1,v2])` | `map()` / `pd.cut()` |

| Correlation | `np.corrcoef([a, b])` | `df.corr()` |
| Percentile | `np.percentile(arr, q)` | `df["col"].quantile(q/100)` |
| Cumulative sum | `np.cumsum(arr)` | `df["col"].cumsum()` |
| Sort indices | `np.argsort(arr)` | `df.sort_values(...).index` |
| Unique values | `np.unique(arr)` | `df["col"].unique()` |
| Reshape | `arr.reshape(r, c)` | — (use before DataFrame creation) |
| Min-Max scale | `(x - x.min()) / (x.max()-x.min())` | Manual or sklearn |
| Z-score | `(x - x.mean()) / x.std()` | Manual or sklearn |
