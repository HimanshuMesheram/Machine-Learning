### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.


```python
# 1. Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = student['marks'].quantile(0.25)
Q3 = student['marks'].quantile(0.75)

# 2. Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# 3. Define the upper and lower bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 4. Filter the DataFrame to find the outliers
outliers = student[(student['marks'] < lower_bound) | (student['marks'] > upper_bound)]
print(outliers)
```

### Q100. Create an end-to-end analysis workflow.
> 💡 **Hint:** Follow this pipeline:  
> 1. Load data → 2. Inspect (shape, dtypes, nulls) → 3. Clean (fillna, drop_duplicates) → 4. Feature engineer → 5. GroupBy/pivot analysis → 6. Sort and rank → 7. Export results.


```python
student.to_csv('Datasets\student.csv')
```

```python
# ==========================================
# 1. LOAD DATA
# ==========================================
# Read the raw source data into a DataFrame
df = pd.read_csv("Datasets\student.csv")

# ==========================================
# 2. INSPECT DATA
# ==========================================
# Check dimensions, column profiles, and integrity counters
print("Dataset Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nMissing Values Count:\n", df.isnull().sum())

# ==========================================
# 3. CLEAN DATA
# ==========================================
# Standardise text casing to ensure reliable groupings
df["name"] = df["name"].str.strip().str.title()
df["country"] = df["country"].str.strip().str.title()

# Rectify missing fields using column-specific strategies
df["iq"] = df["iq"].fillna(df["iq"].median())
df["marks"] = df["marks"].fillna(df["marks"].mean())
df["LPA"] = df["LPA"].fillna(0)

# Purge redundant record rows from the system
df = df.drop_duplicates(keep="first")

# ==========================================
# 4. FEATURE ENGINEER
# ==========================================
# Create advanced derived metrics for deep categorical analysis
df["total_score"] = df["iq"] + df["marks"]
df["efficiency_ratio"] = (df["LPA"] / df["total_score"]).round(4)

# ==========================================
# 5. GROUPBY / PIVOT ANALYSIS
# ==========================================
# Agglomerate metrics into structured matrices with margins
summary_pivot = pd.pivot_table(
    df,
    values=["LPA", "total_score"],
    index="country",
    aggfunc={"LPA": "mean", "total_score": "max"},
    margins=True,
    margins_name="Global Baseline",
    fill_value=0,
)

# ==========================================
# 6. SORT AND RANK
# ==========================================
# Isolate operational rows from summary blocks to perform rankings
analysis_data = summary_pivot.drop("Global Baseline", errors="ignore")

# Sort records based on financial returns
sorted_analysis = analysis_data.sort_values(by="LPA", ascending=False)

# Assign global performance rankings
sorted_analysis["regional_rank"] = (
    sorted_analysis["LPA"].rank(ascending=False, method="dense").astype(int)
)

# Final formatting clean-up
final_report = sorted_analysis.round(2)

# ==========================================
# 7. EXPORT RESULTS
# ==========================================
# Deliver formatted documents to local storage arrays
final_report.to_csv("regional_student_analysis.csv", index=True)
# final_report.to_excel("regional_student_analysis.xlsx", index=True)

print("\n--- Pipeline Completed: Final Analytical Matrix ---")
print(final_report)
```
